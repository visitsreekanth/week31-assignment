### Fill-Mask

- **Task performed:** Predicts possible words or tokens that can replace a special mask token such as `<MASK>`, ranking the suggestions by model score.
- **NLP application represents:** Text completion, language understanding demonstrations, assisting users with writing and learning contextual word relationships.

A masked-language model uses the surrounding context to predict the missing token. The standard fill-mask task expects one mask token in each input.

Named Entity Recognition (NER)

- **Task performed:** Finds named entities in text and assigns categories such as person (PER), organisation (ORG) and location (LOC). As group_entities=True is deprecated in Hugging Face Transformers, With **aggregation_strategy="simple"**, related subword predictions are grouped into entity spans.
- **NLP application represents:** Extracting names and organisations from documents, analysing news articles, processing CVs and supporting information extraction.

The exact entity labels depend on the pretrained model. Not every NER model supports the same label set.

# **Question-answering**

```python
from transformers import pipeline
'''
Text pipelines that should just be LLMs
question-answering and Text2TextGenerationPipeline, including its related SummarizationPipeline and TranslationPipeline, were deprecated and will now be removed.
'''
# Using text-generation pipeline with an instruction/chat model as required in transformers v5
generator = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B-Instruct")

question = "Where do I work?"
context = "My name is Sylvain and I work at Hugging Face in Brooklyn."

# Format the prompt using a chat template
messages = [
    {"role": "system", "content": "Answer the question based strictly on the provided context."},
    {"role": "user", "content": f"Context: {context}\n\nQuestion: {question}"}
    ]


# Generate the answer
outputs = generator(messages, max_new_tokens=20, return_full_text=False)
qa_result = outputs[0]["generated_text"]

print("Context:", context)
print("Question:", question)
print("Answer:", qa_result)
```

Question Answering

- **Task performed:** Uses a question and a supporting context to identify the text span that best answers the question. This example uses **text-generation** for question answering as question-answering and Text2TextGenerationPipeline, including its related SummarizationPipeline and TranslationPipeline, were deprecated.
- **NLP application represents:** FAQ systems, searching knowledge-base articles, document analysis and answering questions about policies or manuals.

The answer is based on the supplied context. This pipeline should not be assumed to know facts that are absent from that context.

## Part B – Tokenizer and Sequence Classification

- AutoTokenizer: loads the tokenizer associated with a checkpoint.
- AutoModel: loads the base Transformer model and returns contextual hidden representations.
- AutoModelForSequenceClassification: loads a Transformer with a classification head for predicting labels.

The sentiment checkpoint below is fine-tuned for positive/negative English sentiment classification.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

checkpoint = "distilbert-base-uncased-finetuned-sst-2-english"

# TODO
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModelForSequenceClassification.from_pretrained(checkpoint)

text = "This course is excellent."

inputs = tokenizer(text, return_tensors="pt")

print(inputs)
input_ids = inputs["input_ids"]
print("tokens:", tokenizer.convert_ids_to_tokens(input_ids[0]))
attention_mask = inputs["attention_mask"]
print("Decoded text:", tokenizer.decode(input_ids[0], skip_special_tokens=False))
print("Decoded text without special tokens:", tokenizer.decode(input_ids[0], skip_special_tokens=True))


outputs = model(**inputs)

print(outputs.logits)

# TODO: Convert logits to probabilities
probabilities = torch.softmax(outputs.logits, dim=1)
predicted_class_ids = torch.argmax(probabilities, dim=1)

print("Logits:\n", outputs.logits)
print("\nProbabilities:\n", probabilities)

id2label = model.config.id2label
for text, probs, class_id in zip(
    text, probabilities, predicted_class_ids
):
    label = id2label[int(class_id)]
    confidence = float(probs[class_id])
    print(f"Text: {text}")
    print(f"Predicted label: {label}")
    print(f"Confidence: {confidence:.4f}\n")
```

## Reflection

Answer:

1. Why is tokenization necessary?

2. What are the purposes of `[CLS]` and `[SEP]`?

3. What are logits?

4. Why is Softmax applied?


- **input_ids:** Numerical vocabulary IDs representing the tokens. Neural networks process numbers rather than raw words.
- **tokens:** The readable token pieces corresponding to those IDs. A word may be split into more than one subword token.
- **attention_mask:** Indicates which positions contain real input tokens (1) and which positions are padding (0). It helps the model ignore padding.
- **Decoding:** Converts token IDs back into readable text. Special tokens can be retained or skipped.

**1. Why is tokenisation necessary?**

Tokenisation breaks text into tokens and maps them to numerical IDs in the model's vocabulary. Transformer models cannot directly process ordinary sentences; they require numerical inputs. Tokenisation also adds required special tokens, creates attention masks and can split unfamiliar words into subword units.

**2. What are the purposes of `[CLS]` and `[SEP]`?**

[CLS] is a special token at the beginning of a BERT-style input. Its final contextual representation is commonly used for sequence-level classification.
[SEP] marks the end of a sequence or separates two input sequences. These tokens help the model interpret the structure of its input.

**3. What are logits?**

Logits are the raw scores produced by the model's classification layer before probability conversion. These are not probabilities and are not restricted to values between zero and one.

**4. Why is Softmax applied?**

Softmax converts the logits into normalised values between zero and one that sum to one across the classes. This makes it easier to interpret and compare the model's estimated probability for each class. The class with the largest probability is selected as the prediction. In simple words, Softmax is applied to convert the model's raw logits into normalized probability values for each class, making the model's prediction easier to interpret.

For this checkpoint, the classes are normally `NEGATIVE` and `POSITIVE`. The probability is the model's estimated confidence, not proof that the sentiment interpretation is correct.

## Part C – Transformer Architectures

Answer:

- Which tasks are suitable for encoder-only models?
- Which tasks are suitable for decoder-only models?
- Which tasks require encoder-decoder models?
- Explain the role of attention.


**1. Why is BERT considered an encoder model?**

BERT is an encoder-only Transformer. Its encoder layers use self-attention to build contextual representations by considering both the left and right context of a token (bidirectional context). This makes BERT effective for understanding and analysing input text, including sentiment analysis, classification, named entity recognition and extractive question answering.

**2. Why is GPT considered a decoder model?**

GPT is a decoder-only Transformer. It uses causal (masked) self-attention, so each token can attend to earlier tokens but not future tokens. It predicts the next token based on the preceding context and repeats this process to generate text. This makes GPT suitable for text generation and conversational applications.

**3. Which architecture would you choose?**

- **Sentiment analysis:** Encoder-only (for example, BERT), because the task requires understanding the whole input and assigning a class label.
- **Text generation:** Decoder-only (for example, GPT), because it generates a sequence one token at a time using previous tokens as context.
- **Translation:** Encoder-decoder (for example, the Transformer architecture used by models such as T5 or the original sequence-to-sequence Transformer). The encoder represents the source sentence, while the decoder generates the translated target sentence while attending to the encoded source.

These are common architectural choices, not absolute restrictions. Other model architectures can also be adapted to some of these tasks.

**4. Explain the role of attention.**

Attention allows a Transformer to determine which other tokens are important when processing a particular token. Self-attention calculates relationships between tokens in the same input sequence and uses these relationships to build contextual representations. For example, attention can help a model connect a pronoun to the noun it refers to, even when other words appear in between. Multi-head attention performs this process through multiple learned attention patterns, allowing the model to capture different relationships. Attention helps Transformers process context and model long-range dependencies more effectively than approaches that process text strictly one word at a time.

## Part D – Critical Evaluation (500 words)

Discuss:

- strengths of pretrained LLMs
- limitations
- hallucinations
- bias
- shallow understanding
- resource requirements
- when fine-tuning is preferred
- ethical considerations


Pretrained Large Language Models (**LLMs**) are useful for organisations that want to use Natural Language Processing (NLP) without developing a model from scratch. These models are trained using large amounts of text, so they already have a good understanding of language patterns, grammar and the relationships between words.
Using platforms such as **Hugging Face**, organisations can use pretrained models for tasks such as sentiment analysis, Named Entity Recognition (NER), question answering (Deprecated however writing here as mentioned in videos) and text generation. This can save development time, reduce the amount of training data required and make it easier to build an initial NLP solution. Transfer learning can also be used to adapt an existing model for a specific task.

**Limitations:**
Pretrained models also have some limitations. Their performance depends on the data used to train them. A model may work well with general English but may not perform as well with specialised areas such as finance, healthcare, law or technical subjects. Models can also have difficulties with ambiguous sentences, complex reasoning and domain-specific terminology. Therefore, organisations should test the model using examples that are similar to their actual business requirements instead of assuming that the model will always give correct results. Organizations can use Retrieval Augmented Generation (RAG) and Model Context Protocol to model get the organization latest/current data.

**Hallucination:**
One of the main important limitations of LLM is hallucination. Sometimes LLM can produce answer that sounds correct and confident but it is wrong with the available information. This can be a serious problem when AI is used for business decisions or any other important decisions. Using approved knowledge sources, good prompts, validation rules and human review can help reduce hallucinations, although they cannot completely eliminate the problem.

**Bias:**
Bias is another concern with pretrained models. As models learn from large amounts of existing data, they can also learn biases or stereotypes present in that training data. This may result in different outputs for certain people. Organisations should therefore test their models with different types of users and monitor the results.

**shallow understanding:**
LLMs can have shallow understanding, meaning that they may generate convincing answers based on learned patterns without actually understanding the meaning or real-world situation. This can become a problem when a task requires detailed reasoning or accurate knowledge of a specific situation.

**Resource Requirements:**
The main problem of current AI industry is resource and cost requirements. Large models can require significant memory, processing power, GPUs and power/water. This can become expensive when the model is used by millions/billions of users or needs to be fine-tuned. Smaller models (Knowledge distillation) may be cheaper and faster but there can be a trade-off between cost, speed and accuracy. Organisations need to consider the cost of hosting, security, maintenance and monitoring.

**When fine-tuning would be preferred:**
Fine-tuning can be considered when a pretrained model does not provide the required performance for a specific task and the organisation has enough suitable training data. For example, company could fine-tune a classification model using its own customer-support messages so that it can better identify business-specific categories. Fine-tuning can improve the model's performance for a particular task, but it requires good-quality data, computing resources, testing and ongoing monitoring. If the main requirement is to give the model better performance or provide with current/latest company information, prompt engineering or Retrieval Augmented Generation (RAG) may be a better option than fine-tuning.

**Ethical considerations:**
Ethical considerations are important before deploying LLM in any organization. Companies need to protect personal and confidential information to use data appropriately, consider bias and provide human oversight where decisions may have significant impact. There should be responsible teams to validate the AI generated outputs and monitor the system after deployment. Overall, pretrained LLMs provide a useful starting point for NLP applications but organisations should choose between direct use, prompt engineering, RAG and fine-tuning based on the required accuracy, business needs, risks, cost and available resources.
